```python
import os
from dotenv import load_dotenv
from dotenv import dotenv_values
load_dotenv()
from telethon import TelegramClient


api_id = os.environ["api_id"]
api_hash = os.environ["api_hash"]
client = TelegramClient('anon', api_id, api_hash)
```

```python
import pandas as pd
```

```python
messages_data = []
async def scrape_channel_messages(channel_name):
    async with client:
        
        channel = await client.get_entity(channel_name)

        # Retrieve the last 100 messages from the channel
        async for message in client.iter_messages(channel, limit=1000):
            # Get message date and time
            message_date = message.date.strftime('%Y-%m-%d')  # Format as YYYY-MM-DD
            message_time = message.date.strftime('%H:%M:%S')  # Format as HH:MM:SS

            # Generate a link to the message (not important in most cases)
            if message.id:
                message_link = f"https://t.me/{channel.username}/{message.id}"
            else:
                message_link = None

            # Get the full text of the message 
            message_text = message.text if message.text else "No Text"

            # Append the data as a dictionary to the messages_data list
            
            messages_data.append({
                'Channel Name': channel.title,
                'Date': message_date,
                'Time': message_time,
                'Link': message_link,
                'Message': message_text
            })

# Run the scraping and display the table
import nest_asyncio
nest_asyncio.apply()  # Allow nested event loops in Jupyter or similar environments

await scrape_channel_messages('https://t.me/argentinacasinos') #***** is the name of the channel. to scrape multiple channels pass the list and put this line into a loop where **** will be in an f-string


df = pd.DataFrame(messages_data)
print(df)


df.to_csv("tg_ac.csv") #can also be an f-string
```
